```python
import joblib
from feature_engineering import FeatureExtractor

# Load the trained model
model = joblib.load("../models/xgboost_url_detector.pkl")

# Create extractor
extractor = FeatureExtractor()

# Extract features
features = extractor.extract_features("https://www.google.com")

# Predict
prediction = model.predict(features)

print(prediction)
```

```python
test_urls = [
    "https://www.google.com",
    "https://github.com",
    "https://facebook.com",
    "http://paypal-login-security.xyz/update",
    "https://amazon.verify-user-security.com/login?id=12345"
]

for url in test_urls:
    features = extractor.extract_features(url)
    prediction = model.predict(features)[0]

    if prediction == 0:
        print(f"{url} --> Legitimate")
    else:
        print(f"{url} --> Phishing")
```

```python
url = "http://paypal-login-security.xyz/update"

features = extractor.extract_features(url)

prediction = model.predict(features)[0]
probability = model.predict_proba(features)[0]

print("Prediction :", prediction)
print("Probability:", probability)
```
